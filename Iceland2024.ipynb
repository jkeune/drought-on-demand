```python
aid_spi1   = {}
aid_spi12  = {}
aid_spei12 = {}

for severity in ['mild','moderate','severe','extreme','any']:

    if severity == "mild":
        thresh_min = -1
        thresh_max = 0
    if severity == "moderate":
        thresh_min = -1.5
        thresh_max = -1
    if severity == "severe":
        thresh_min = -2
        thresh_max = -1.5
    if severity == "extreme":
        thresh_min = -999
        thresh_max = -2
    if severity == "any":
        thresh_min = -999
        thresh_max = 0
        
    # SPI-1
    rspi   = spi1*countrymask_3d_xr
    dmask  = xr.where((rspi > thresh_min) & (rspi <= thresh_max), 1, np.nan)
    tmp    = 100 * (dmask*area_3d).sum(dim=["lat", "lon"]) / area_3d.sum(dim=["lat", "lon"])
    aid_spi1[severity] = tmp
    del(tmp, rspi, dmask)
    
    # SPI-12
    rspi   = spi12*countrymask_3d_xr
    dmask  = xr.where((rspi > thresh_min) & (rspi <= thresh_max), 1, np.nan)
    tmp    = 100 * (dmask*area_3d).sum(dim=["lat", "lon"]) / area_3d.sum(dim=["lat", "lon"])
    aid_spi12[severity] = tmp
    del(tmp, rspi, dmask)

    # SPEI-12
    rspi   = spei12*countrymask_3d_xr
    dmask  = xr.where((rspi > thresh_min) & (rspi <= thresh_max), 1, np.nan)
    tmp    = 100 * (dmask*area_3d).sum(dim=["lat", "lon"]) / area_3d.sum(dim=["lat", "lon"])
    aid_spei12[severity] = tmp
    del(tmp, rspi, dmask)
```

```python
aid_spi1    = {}
aid_spi12   = {}
aid_spei12  = {}

for severity in ['mild','moderate','severe','extreme','any']:

    if severity == "mild":
        thresh_min = -1
        thresh_max = 0
    if severity == "moderate":
        thresh_min = -1.5
        thresh_max = -1
    if severity == "severe":
        thresh_min = -2
        thresh_max = -1.5
    if severity == "extreme":
        thresh_min = -999
        thresh_max = -2
    if severity == "any":
        thresh_min = -999
        thresh_max = 0
        
    # SPI-1
    rspi   = spi1*nutsmask_3d_xr
    dmask  = xr.where((rspi > thresh_min) & (rspi <= thresh_max), 1, np.nan)
    tmp    = 100 * (dmask*area_3d).sum(dim=["lat", "lon"]) / area_3d.sum(dim=["lat", "lon"])
    aid_spi1[severity] = tmp
    del(tmp, rspi, dmask)

    # SPI-12
    rspi   = spi12*nutsmask_3d_xr
    dmask  = xr.where((rspi > thresh_min) & (rspi <= thresh_max), 1, np.nan)
    tmp    = 100 * (dmask*area_3d).sum(dim=["lat", "lon"]) / area_3d.sum(dim=["lat", "lon"])
    aid_spi12[severity] = tmp
    del(tmp, rspi, dmask)

    # SPEI-12
    rspi   = spei12*nutsmask_3d_xr
    dmask  = xr.where((rspi > thresh_min) & (rspi <= thresh_max), 1, np.nan)
    tmp    = 100 * (dmask*area_3d).sum(dim=["lat", "lon"]) / area_3d.sum(dim=["lat", "lon"])
    aid_spei12[severity] = tmp
    del(tmp, rspi, dmask)
```
